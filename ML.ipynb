# 梯度下降（Gradient Descent）完整数学知识与公式
梯度下降是机器学习中最核心的**优化算法**，核心目标是通过迭代更新模型参数（如线性回归中的 \( w \) 和 \( b \)），最小化成本函数 \( J(w,b) \)。以下是从基础原理到公式推导、变体的完整数学内容。

## 一、梯度下降核心目标
梯度下降的本质是：**沿着成本函数 \( J(w,b) \) 梯度的反方向，逐步更新参数，最终找到成本函数的最小值点**。

核心目标数学表达：
找到最优参数 \( (w^*, b^*) \)，使得 \( J(w^*, b^*) = \min_{w,b} J(w,b) \)

其中 \( J(w,b) \) 是线性回归场景的**均方误差成本函数**：
$J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)^2$

- \( f_{w,b}(x^{(i)}) = w \cdot x^{(i)} + b \)（线性回归预测函数）
- \( m \)：训练样本数
- \( x^{(i)} \)：第 \( i \) 个样本的输入特征（如房屋面积）
- \( y^{(i)} \)：第 \( i \) 个样本的真实标签（如房屋真实成本）

## 二、梯度下降基础原理
### 1. 梯度的数学定义
梯度是**多变量函数的偏导数向量**，描述了函数在某点的“变化率最大方向”。对于成本函数 \( J(w,b) \)，梯度记为 \( \nabla J(w,b) \)：
$\nabla J(w,b) = \left( \frac{\partial J(w,b)}{\partial w}, \frac{\partial J(w,b)}{\partial b} \right)$

- \( \frac{\partial J(w,b)}{\partial w} \)：\( J(w,b) \) 对 \( w \) 的偏导数（仅关注 \( w \) 变化时 \( J \) 的变化率）
- \( \frac{\partial J(w,b)}{\partial b} \)：\( J(w,b) \) 对 \( b \) 的偏导数（仅关注 \( b \) 变化时 \( J \) 的变化率）

### 2. 梯度的物理意义
- 梯度的**方向**：成本函数 \( J(w,b) \) 增长最快的方向；
- 梯度的**反方向**：成本函数 \( J(w,b) \) 下降最快的方向（梯度下降的核心依据）；
- 梯度的**大小**：成本函数在该点的变化率大小。

## 三、梯度下降核心更新公式
### 1. 单参数场景（仅 \( w \)，无 \( b \)）
$w := w - \alpha \cdot \frac{\partial J(w)}{\partial w}$

- \( := \)：赋值符号（用右边的值更新左边的参数）
- \( \alpha \)：学习率（步长，控制每次更新的幅度）
- \( \frac{\partial J(w)}{\partial w} \)：\( J(w) \) 对 \( w \) 的偏导数（梯度）

### 2. 双参数场景（\( w \) 和 \( b \)，线性回归核心）
**两个参数需同步更新**：
$\begin{cases}
w := w - \alpha \cdot \frac{\partial J(w,b)}{\partial w} \\
b := b - \alpha \cdot \frac{\partial J(w,b)}{\partial b}
\end{cases}$

#### 符号说明
| 符号 | 数学含义 | 通俗解释 |
| :--- | :--- | :--- |
| \( \alpha \) | 学习率（超参数） | 每次更新的“步长”，比如 \( \alpha=0.01 \) 表示小步慢更 |
| \( \frac{\partial J(w,b)}{\partial w} \) | \( J(w,b) \) 对 \( w \) 的偏导数 | \( w \) 每变化1单位，\( J(w,b) \) 的变化量 |
| \( \frac{\partial J(w,b)}{\partial b} \) | \( J(w,b) \) 对 \( b \) 的偏导数 | \( b \) 每变化1单位，\( J(w,b) \) 的变化量 |

## 四、偏导数（梯度）的数学推导（线性回归场景）
### 1. 第一步：拆分成本函数
$J(w,b) = \frac{1}{2m} \sum_{i=1}^{m} \underbrace{\left( f_{w,b}(x^{(i)}) - y^{(i)} \right)^2}_{误差项：记为 \mathcal{L}(i)}$

### 2. 第二步：推导 \( \frac{\partial J(w,b)}{\partial w} \)
根据偏导数线性性质：
$\frac{\partial J(w,b)}{\partial w} = \frac{1}{2m} \sum_{i=1}^{m} \frac{\partial \mathcal{L}(i)}{\partial w}$

应用链式法则（\( $$ \frac{\partial (u^2)}{\partial w} = 2u \cdot \frac{\partial u}{\partial w}  $$ )）：
$\frac{\partial \mathcal{L}(i)}{\partial w} = 2 \cdot \left( f_{w,b}(x^{(i)}) - y^{(i)} \right) \cdot x^{(i)}$

化简（2和1/2抵消）：
$\frac{\partial J(w,b)}{\partial w} = \frac{1}{m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right) \cdot x^{(i)}$

### 3. 第三步：推导 \( \frac{\partial J(w,b)}{\partial b} \)
同理，对 \( b \) 求偏导（\( f_{w,b}(x^{(i)}) \) 对 \( b \) 的偏导数为1）：
$\frac{\partial J(w,b)}{\partial b} = \frac{1}{m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)$

### 4. 最终更新公式
$\begin{cases}
w := w - \alpha \cdot \frac{1}{m} \sum_{i=1}^{m} \left( w \cdot x^{(i)} + b - y^{(i)} \right) \cdot x^{(i)} \\
b := b - \alpha \cdot \frac{1}{m} \sum_{i=1}^{m} \left( w \cdot x^{(i)} + b - y^{(i)} \right)
\end{cases}$

## 五、学习率 \( \alpha \) 的数学意义与选择
### 1. 数学意义
$参数更新量 = \alpha \cdot 梯度$

- 若 \( \alpha \) 过大：步长太长，可能跳过最小值，\( J(w,b) \) 震荡不收敛；
- 若 \( \alpha \) 过小：步长太短，收敛速度极慢；
- 若 \( \alpha \) 合适：\( J(w,b) \) 单调递减，逐步逼近最小值。

### 2. 常用取值范围
$\alpha \in \{0.001, 0.01, 0.1, 0.3\}$（需根据数据缩放调整）

## 六、梯度下降的三种变体
### 1. 批量梯度下降（Batch Gradient Descent, BGD）
#### 公式
$\begin{cases}
w := w - \alpha \cdot \frac{1}{m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right) x^{(i)} \\
b := b - \alpha \cdot \frac{1}{m} \sum_{i=1}^{m} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)
\end{cases}$

#### 特点
- 用全部 \( m \) 个样本计算梯度，梯度准确，\( J(w,b) \) 单调递减；
- 时间复杂度 \( O(m) \)，样本量大时迭代慢。

### 2. 随机梯度下降（Stochastic Gradient Descent, SGD）
#### 公式
$\begin{cases}
w := w - \alpha \cdot \left( f_{w,b}(x^{(i)}) - y^{(i)} \right) x^{(i)} \\
b := b - \alpha \cdot \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)
\end{cases}$

- \( i \)：每次迭代随机选择的样本索引

#### 特点
- 用1个样本计算梯度，时间复杂度 \( O(1) \)，速度快；
- 梯度有噪声，\( J(w,b) \) 围绕最小值震荡。

### 3. 小批量梯度下降（Mini-Batch Gradient Descent, MBGD）
#### 公式
$\begin{cases}
w := w - \alpha \cdot \frac{1}{b} \sum_{i=k}^{k+b-1} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right) x^{(i)} \\
b := b - \alpha \cdot \frac{1}{b} \sum_{i=k}^{k+b-1} \left( f_{w,b}(x^{(i)}) - y^{(i)} \right)
\end{cases}$

- \( b \)：批量大小（通常 \( b=32, 64, 128 \)）
- \( k \)：批量起始索引

#### 特点
- 平衡准确性和速度，工业界最常用。

## 七、收敛判断的数学标准
1. 成本函数差值小于阈值 \( \epsilon \)（如 \( \epsilon=10^{-6} \)）：
$|J(w_{t+1}, b_{t+1}) - J(w_t, b_t)| < \epsilon$

2. 梯度的模长小于阈值 \( \epsilon \)：
$\|\nabla J(w_t, b_t)\| = \sqrt{\left( \frac{\partial J}{\partial w} \right)^2 + \left( \frac{\partial J}{\partial b} \right)^2} < \epsilon$

## 八、关键数学注意事项
1. **参数同步更新**：先计算所有梯度，再同时更新 \( w \) 和 \( b \)；
2. **数据标准化**：特征尺度差异大时需标准化（\( x' = \frac{x - \mu}{\sigma} \)），否则收敛慢；
3. **凸函数特性**：线性回归的成本函数是凸函数，梯度下降能找到全局最小值。

# 逻辑回归中的正则化（Regularization）详解

在逻辑回归（及所有机器学习模型）中，**正则化（Regularization）** 是解决「过拟合」的核心技术，其本质是：在模型的损失函数中加入「权重惩罚项」，约束模型参数（权重 $ \mathbf{w} $）的大小，让模型在“拟合训练数据”和“保持参数简洁”之间找到平衡，最终提升模型在未知数据上的泛化能力。


## 一、为什么逻辑回归需要正则化？
逻辑回归是**线性模型**，但依然会出现过拟合，尤其是以下情况：
- 特征数量多（甚至超过样本数）；
- 部分特征与标签是“虚假相关”（训练集特有噪声）；
- 模型权重 $ \mathbf{w} $ 数值过大，导致对个别特征过度敏感（比如某个特征的微小变化，会让预测概率大幅波动）。

过拟合的表现：训练集准确率很高，但测试集准确率骤降（模型“死记硬背”了训练数据的噪声，却不会推广到新数据）。

正则化的核心思路：**“惩罚权重过大的模型”**——即使某个权重能让训练损失更小，但只要它太大，就通过惩罚项增加总损失，逼迫模型选择更小、更均衡的权重，从而降低模型的“复杂度”，避免过拟合。


## 二、正则化的数学原理：修改损失函数
逻辑回归的原始成本函数（无正则化）是：
$$ J(\mathbf{w}, b) = \frac{1}{m}\sum_{i=1}^m \left[ -y^{(i)}\log h - (1-y^{(i)})\log(1-h) \right] $$
（$ h = h_{\mathbf{w}, b}(\mathbf{x}^{(i)}) $ 是预测概率）

加入正则化后，新的成本函数为：
$$ J_{\text{reg}}(\mathbf{w}, b) = J(\mathbf{w}, b) + \lambda \cdot R(\mathbf{w}) $$
- $ R(\mathbf{w}) $：**正则化项（惩罚项）**，仅惩罚权重 $ \mathbf{w} $（不惩罚偏置 $ b $，因为偏置是常数项，不会导致过拟合）；
- $ \lambda $：**正则化强度参数**（超参数）：
  - $ \lambda = 0 $：无正则化，退化为原始损失函数；
  - $ \lambda $ 越大：惩罚越强，权重被约束得越小，可能导致欠拟合（模型太简单，连训练数据的规律都拟合不好）；
  - $ \lambda $ 越小：惩罚越弱，越容易过拟合。


## 三、逻辑回归中3种常用正则化类型
### 1. L2正则化（Ridge Regression 脊回归）【默认首选】
#### 公式
正则化项为权重的「平方和」：
$$ R(\mathbf{w}) = \frac{1}{2}\sum_{j=1}^n w_j^2 = \frac{1}{2}\|\mathbf{w}\|_2^2 $$
（$ \|\mathbf{w}\|_2 $ 是权重向量的L2范数，加 $ 1/2 $ 是为了求导后抵消系数，简化计算）

加入L2正则后的成本函数：
$$ J_{\text{reg}} = \frac{1}{m}\sum_{i=1}^m L(h, y^{(i)}) + \frac{\lambda}{2m}\sum_{j=1}^n w_j^2 $$
（注：部分教材会把 $ \lambda $ 除以 $ m $，让正则项与样本数无关，更稳定）

#### 核心特点
- 「权重衰减（Weight Decay）」：L2正则会让所有权重 $ w_j $ 趋向于 **$ 0 $ 附近的小值**，但不会让任何权重完全归零；
- 效果：模型权重分布更均衡，不会过度依赖某几个特征，有效抑制过拟合；
- 适用场景：绝大多数情况（尤其是特征都有一定价值，无需删除特征时）；
- Sklearn参数：`penalty='l2'`（默认），$ C=1/\lambda $（$ C $ 越大，$ \lambda $ 越小，正则化越弱）。

#### 梯度更新变化
之前推导的权重梯度为 $ \frac{\partial J}{\partial w_j} = \frac{1}{m}\sum (h-y)x_j^{(i)} $，加入L2正则后：
$$ \frac{\partial J_{\text{reg}}}{\partial w_j} = \frac{1}{m}\sum (h-y)x_j^{(i)} + \frac{\lambda}{m}w_j $$
更新规则变为：
$$ w_j = w_j - \alpha \left( \frac{1}{m}\sum (h-y)x_j^{(i)} + \frac{\lambda}{m}w_j \right) $$
可见，每次更新都会额外减去 $ \alpha \cdot \frac{\lambda}{m}w_j $，让权重“衰减”变小。

### 2. L1正则化（Lasso Regression 拉索回归）
#### 公式
正则化项为权重的「绝对值和」：
$$ R(\mathbf{w}) = \sum_{j=1}^n |w_j| = \|\mathbf{w}\|_1 $$

加入L1正则后的成本函数：
$$ J_{\text{reg}} = \frac{1}{m}\sum_{i=1}^m L(h, y^{(i)}) + \frac{\lambda}{m}\sum_{j=1}^n |w_j| $$

#### 核心特点
- 「稀疏性（Sparsity）」：L1正则会让 **部分权重完全归零**（相当于“删除”了对应的特征），实现「特征选择」；
- 效果：不仅抑制过拟合，还能简化模型（保留重要特征，剔除冗余/无关特征）；
- 适用场景：特征数量多、存在冗余特征（比如100个特征中只有10个有用），想降低模型复杂度时；
- Sklearn参数：`penalty='l1'`，注意：仅支持 `solver='liblinear'` 或 `solver='saga'`（其他求解器不支持L1正则）。

#### 为什么L1能让权重归零？
L2正则的惩罚项是平方函数，导数随权重增大而增大（对大权重惩罚强，但小权重惩罚弱，不会归零）；  
L1正则的惩罚项是绝对值函数，导数是常数（对所有非零权重的惩罚强度相同）——当某个特征不重要时，模型会倾向于让其权重归零，彻底放弃该特征，从而实现稀疏化。

### 3. 弹性网（ElasticNet）【L1+L2结合】
#### 公式
同时加入L1和L2正则项，兼顾两者优点：
$$ R(\mathbf{w}) = \alpha\sum_{j=1}^n |w_j| + \frac{1-\alpha}{2}\sum_{j=1}^n w_j^2 $$
- $ \alpha \in [0,1] $：控制L1和L2的比例（$ \alpha=1 $ 等价于L1，$ \alpha=0 $ 等价于L2）。

加入弹性网后的成本函数：
$$ J_{\text{reg}} = \frac{1}{m}\sum_{i=1}^m L(h, y^{(i)}) + \frac{\lambda}{m}\left( \alpha\sum |w_j| + \frac{1-\alpha}{2}\sum w_j^2 \right) $$

#### 核心特点
- 解决L1的缺陷：L1在特征高度相关时，可能随机删除其中一个；弹性网会保留相关特征（通过L2的权重衰减），同时通过L1实现特征选择；
- 适用场景：特征多、存在多重共线性（特征相关）时；
- Sklearn参数：`penalty='elasticnet'`，需指定 `l1_ratio=α`（比如 `l1_ratio=0.5` 表示L1和L2各占一半），且仅支持 `solver='saga'`。


## 四、正则化的关键注意事项
### 1. 正则化只惩罚权重 $ \mathbf{w} $，不惩罚偏置 $ b $
偏置 $ b $ 是常数项（对应 $ x_0=1 $ 的特征），其数值大小不会导致模型对特征敏感，惩罚 $ b $ 反而会影响模型的拟合能力，因此所有正则化项都只针对权重 $ \mathbf{w} $。

### 2. 特征放缩是正则化的“前提”
正则化对特征尺度敏感！比如特征A（$ 0 \sim 1000 $）和特征B（$ 0 \sim 1 $），若未放缩，模型会认为特征A的权重 $ w_A $ 数值大是“正常的”，正则化会过度惩罚 $ w_A $，导致模型忽略特征A的真实价值。

因此，**使用正则化前必须做特征放缩**（优先标准化Z-Score），让所有特征尺度一致，确保正则化对每个权重的惩罚公平。

### 3. 正则化强度 $ \lambda $ 的选择
- 不能太大：否则权重被过度压制，模型太简单，欠拟合；
- 不能太小：否则惩罚无效，依然过拟合；
- 最佳选择：通过「交叉验证（Cross Validation）」，比如使用Sklearn的 `LogisticRegressionCV` 自动搜索最优 $ \lambda $（对应最优 $ C=1/\lambda $）。

### 4. 与梯度下降的兼容性
- L2正则：损失函数是凸函数，梯度光滑，所有求解器（lbfgs、sag、liblinear）都支持，收敛稳定；
- L1正则：损失函数在权重为 $ 0 $ 处不可导（导数突变），仅支持liblinear（坐标下降法）、saga（梯度下降变种）等求解器；
- 弹性网：仅支持saga求解器。


## 五、总结
正则化的核心是「通过惩罚权重大小，平衡拟合能力与泛化能力，抑制过拟合」：

| 正则化类型 | 核心作用                | 特点                  | Sklearn参数                          |
|------------|-------------------------|-----------------------|-------------------------------------|
| L2（默认） | 权重衰减，抑制过拟合    | 权重非稀疏，保留所有特征 | `penalty='l2'`，$ C=1/\lambda $             |
| L1         | 特征选择+抑制过拟合     | 权重稀疏，删除冗余特征 | `penalty='l1'`，`solver='liblinear'`|
| 弹性网     | 兼顾特征选择与权重均衡  | 解决多重共线性        | `penalty='elasticnet'`，`l1_ratio=α`|

实际使用时：
1. 优先用L2正则（默认配置），配合标准化；
2. 特征冗余多时，用L1正则或弹性网；
3. 用 `LogisticRegressionCV` 自动选择最优正则化强度 $ C $，避免手动调参的盲目性。
