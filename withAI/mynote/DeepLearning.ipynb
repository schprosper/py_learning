MLP 是 Multilayer Perceptron 的缩写，中文译为「多层感知机」，它是一种最基础、最经典的前馈神经网络Feedforward Neural Network
它是单层感知机（Single-Layer Perceptron）的升级版本（单层感知机只能解决线性可分问题，比如简单的「与 / 或」逻辑，无法解决「异或」这类非线性问题）  
二、MLP 的核心结构（三层基本组成）
MLP 的结构非常直观，由「输入层→隐藏层→输出层」三层神经元（Neuron）组成，相邻层的神经元全连接（每个神经元与下一层所有神经元相连）：
1. 输入层（Input Layer）
作用：接收原始数据或特征向量，不进行任何计算，仅将数据传递给下一层；
神经元数量 = 输入数据的特征维度（例如：输入是 28×28 的手写数字图像，展平后是 784 维向量，输入层就有 784 个神经元）。
2. 隐藏层（Hidden Layer）
作用：核心的「特征提取器」—— 通过神经元的加权求和 + 激活函数的非线性转换，将输入层的原始特征映射为更高维度、更具区分度的抽象特征；
数量：可以是 1 层（称为「浅层 MLP」）或多层（称为「深层 MLP」，即早期的深度学习模型）；
神经元数量：无固定规则，通常根据任务复杂度调整（越多拟合能力越强，但易过拟合）。
3. 输出层（Output Layer）
作用：将隐藏层的抽象特征映射为最终的预测结果；
神经元数量 = 任务目标维度：
分类任务：等于类别数（例如二分类用 1 个神经元，多分类用 N 个神经元，输出为各类别的概率，常用 Softmax 激活）；
回归任务：通常用 1 个神经元（输出连续数值，无需激活函数或用 Linear 激活）。
关键组件：激活函数（Activation Function）
核心作用：打破模型的线性限制—— 如果没有激活函数，无论多少层神经元，整体都等价于一个线性模型，无法拟合非线性问题；
常用激活函数：
隐藏层：ReLU（最常用，解决梯度消失）、Tanh、Sigmoid；
输出层：Softmax（多分类）、Sigmoid（二分类）、Linear（回归）。

三、MLP 的工作原理（两大核心步骤）
1. 前向传播（Forward Propagation）
信号从输入层→隐藏层→输出层单向流动，计算每一层的输出：
对每个神经元：先计算前一层所有神经元输出的「加权和 + 偏置项」（即 z = w·x + b，w 是权重，b 是偏置）；
再将 z 代入激活函数，得到该神经元的输出（即 a = σ(z)，σ 是激活函数）。
最终输出层的结果即为模型的「预测值」。
2. 反向传播（Backward Propagation）
作用：优化模型参数（w 和 b），最小化预测值与真实值的误差（损失函数，如交叉熵、MSE）；
核心逻辑：利用「梯度下降法」，从输出层反向计算损失函数对每个参数的梯度（链式法则），并根据梯度更新参数（w = w - η·∇L(w)，η 是学习率）；
这是 MLP 能够「学习」的核心 —— 通过不断迭代，调整权重和偏置，让模型的预测越来越接近真实值。

四、MLP 的特点与适用场景  
优点：  
结构简单、易于理解和实现；  
非线性拟合能力强，可处理各类非线性任务；   
端到端训练，无需手动设计特征（区别于传统机器学习如 SVM、决策树）。   
缺点：   
全连接结构导致参数过多（例如输入层 1000 个神经元、隐藏层 1000 个神经元，仅这一层的权重就有 100 万个），易过拟合、训练效率低；  
对高维稀疏数据（如文本）或具有空间 / 时序结构的数据（如图像、语音），特征提取效率远低于 CNN（空间结构）、RNN（时序结构）等专用模型。  
适用场景：  
简单的分类 / 回归任务（如 Iris 分类、房价预测）；  
数据维度不高、无明显结构的任务（如表格数据挖掘）；  
作为复杂模型的子模块（如 Transformer 中的 Feed-Forward Network 本质就是两层 MLP）。  
五、MLP 与其他模型的区别  
与单层感知机：MLP 多了隐藏层和激活函数，能解决非线性问题；  
与 CNN：CNN 引入「卷积层 + 池化层」，利用局部感受野和权值共享，更适合处理图像等具有空间结构的数据，参数远少于 MLP；  
与 RNN：RNN 引入「循环结构」，能处理文本、语音等时序数据，捕捉序列依赖关系；  
与深度学习：深层 MLP（多层隐藏层）就是早期的深度学习模型，后续的 CNN、Transformer 都是在 MLP 基础上针对特定数据结构优化而来。
